```python
import urllib.request

text = urllib.request.urlopen("https://drive.google.com/uc?export=download&id=1gNDt93QaPa73a4OQhpEieofD4KTUw0B9").read().decode("utf-8")
text[:131]
```

```python
print(text[2:8]) # extract characters from 2-8
print(text[:59]) # extract characters until the 59th character
print(text[-30:]) # extract the last 30 characters
print(text[::8]) # extract every eighth character
print(text[12:20]) # extract a word from the text
print("The amount of characters: ", len(text)) # print the amount of characters
```

```python
words = text.split(" ") # split the sentences into separate words
print(words)

print(*words[0:10]) # extract the first ten words
print(*words[-8:]) # extract the last eight words
print("The number of words: ", len(words)) # print the number of words
```
